```python
import heapq

GOAL = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)  # 0 represents the blank space
)

START = (
    (1, 2, 3),
    (4, 0, 6),
    (7, 5, 8)
)

# Possible moves: (row_change, col_change, description)
DIRECTIONS = [
    (-1, 0, "Up"),
    (1, 0, "Down"),
    (0, -1, "Left"),
    (0, 1, "Right")
]

def heuristic(state):
    """Calculate the Manhattan distance for the given puzzle state."""
    total = 0
    for row in range(3):
        for col in range(3):
            value = state[row][col]
            if value != 0:
                goal_row, goal_col = divmod(value - 1, 3)
                total += abs(goal_row - row) + abs(goal_col - col)
    return total

def locate_blank(state):
    """Find the position of the blank (0) in the state."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def move_tile(state, r1, c1, r2, c2):
    """Return a new state with two tiles swapped."""
    grid = [list(row) for row in state]
    grid[r1][c1], grid[r2][c2] = grid[r2][c2], grid[r1][c1]
    return tuple(tuple(row) for row in grid)

def a_star_solver(start_state):
    """Run the A* search algorithm to solve the puzzle."""
    frontier = []
    heapq.heappush(frontier, (heuristic(start_state), 0, start_state, []))
    seen = set()

    while frontier:
        f_cost, g_cost, current, path = heapq.heappop(frontier)

        if current == GOAL:
            return path, current

        if current in seen:
            continue
        seen.add(current)

        blank_r, blank_c = locate_blank(current)

        for dr, dc, action in DIRECTIONS:
            new_r, new_c = blank_r + dr, blank_c + dc
            if 0 <= new_r < 3 and 0 <= new_c < 3:
                updated_state = move_tile(current, blank_r, blank_c, new_r, new_c)
                if updated_state not in seen:
                    new_g = g_cost + 1
                    heapq.heappush(
                        frontier,
                        (new_g + heuristic(updated_state), new_g, updated_state, path + [action])
                    )

    return None, None

if __name__ == "__main__":
    steps, final = a_star_solver(START)

    print("Steps to solve:", steps)
    print("Move count:", len(steps))
    print("\nFinal configuration:")
    for row in final:
        print(row)
```
